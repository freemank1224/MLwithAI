# 起步：给校园咖啡摊估计客流

你正帮校园咖啡摊准备明天的纸杯。过去几天记录了上午卖出的杯数；怎样用这组数做一个简单估计？在真正使用模型前，我们先练习变量、列表、函数和简单的误差计算。

**使用方法**：从上到下选择单元格，按 `Shift + Enter` 运行。先读问题并猜答案，再运行代码。尝试修改带有「动手」的代码，重新运行该单元格及其下方的检查。检查报错意味着可以继续调整；读懂图和数字比记住语法更重要。

## 1. 记录数据：变量和列表

过去四天卖出 18、22、20、24 杯。运行前猜一猜：`sales[0]` 是第几个数字？`len(sales)` 是什么？Python 列表从 **0** 开始编号。

```python
sales = [18, 22, 20, 24]
print("第一天:", sales[0])
print("记录天数:", len(sales))
print("总杯数:", sum(sales))
```

**动手**：把第五天的 16 杯加入列表，运行下面的单元格，再运行检查。重复运行不会重复添加，因为先复制了原列表。

```python
five_days = sales.copy()
five_days.append(16)  # 把 16 改成另一个数字试试，检查会告诉你是否符合任务
print(five_days)
```

```python
assert len(five_days) == 5, "应当有五天记录"
assert five_days[-1] == 16, "第五天应当是 16 杯"
assert sales == [18, 22, 20, 24], "不要改变原来的四天记录"
print("检查通过：第五天已加入，原记录保持不变。")
```

## 2. 从数据到预测：定义函数

先猜：如果把这四天的平均杯数当作明天的预测值，会是多少？下面的函数将计算规则起名为 `average`，这样其他数据也能使用同一规则。`return` 把结果交回给调用它的代码。

```python
def average(values):
    return sum(values) / len(values)

forecast = average(sales)
print("预测明天需要的杯数:", forecast)
```

**动手**：试试把 `average(sales)` 换成 `average(five_days)`，比较预测值。再观察下方检查使用另一组数据：函数需要对不同长度的非空列表都正确，而不只是记住这四天的答案。

```python
print("五天平均:", average(five_days))
assert average([2, 4, 9]) == 5, "总和应除以实际天数"
assert average([8]) == 8, "只有一天时平均数就是当天值"
print("检查通过：规则适用于不同天数。")
```

## 3. 比较预测与现实：误差

如果预测 21 杯，实际卖出 24 杯，那么预测误差有多大？我们用绝对误差 `|预测 - 实际|` 表示差距，忽略高估还是低估。实际工作中两种错误的代价可能不同。

**动手**：修改第二行的预测值，看看误差如何变化；之后运行检查。

```python
def absolute_error(predicted, actual):
    return abs(predicted - actual)

predicted = 21
actual = 24
print("预测:", predicted, "实际:", actual, "绝对误差:", absolute_error(predicted, actual))
```

```python
assert absolute_error(21, 24) == 3, "低估 3 杯的绝对误差应为 3"
assert absolute_error(24, 21) == 3, "高估 3 杯的绝对误差也应为 3"
assert absolute_error(10, 10) == 0, "预测准确时误差应为 0"
print("检查通过：绝对误差的三个例子都正确。")
```

## 4. 把步骤连起来

在最后一次练习中，把前面两个函数组合起来：用前四天的平均数预测第五天，再和第五天的真实值比较。先在纸上算一遍，再运行。

```python
training_days = five_days[:4]
test_day = five_days[4]
prediction = average(training_days)
error = absolute_error(prediction, test_day)
print(f"前四天预测：{prediction:.1f} 杯；第五天实际：{test_day} 杯；误差：{error:.1f} 杯")
```

```python
assert prediction == 21, "预测应只使用前四天"
assert error == 5, "第五天实际 16 杯，与预测 21 杯相差 5 杯"
print("检查通过：预测时没有使用要检验的那一天。")
```
